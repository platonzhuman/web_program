**'use strict'**

- Включение строго режима, при котором многие "старые" невидимые ошибки становятся видимыми и о них можно будет узнать в консоли браузера
- Директива должна быть самой первой в скрипте или в функции

* лучше использовать всегда строгий режим

````python
%%html

<html>
    <head>
        <meta charset="utf-8">
        <meta name="viewport" content ="width=device-width, initial-scale=1">
        <title>js</title>
    </head>
    <body>
        <h1>Hello</h1>
        <p>test</p>
        <script>
            let a = 5;
            switch (a) {
                case 5:
                    doSmth();
                    break;
                case 10:
                    domSmthElse();
                    // break;
                default:
                    doDefault();
            };
            console.log(a)
        </script>
    </body>
</html>
````

* это нужно для того, чтобы загружать скрипты фоново чтобы сайт  быстрее грузился  именно скрипты в боди или в хеде
* async  - он параллельно все подгружает а не откладывает на потом
  1. еще async удобен при внедрении рекламных банеров на сайт
  2. лучше использовать в head
