# Веб-программирование — занятие 1
### Основы JavaScript

**Логика занятия:** что такое JavaScript и где он выполняется → из чего состоит клиентский JS → что он умеет и чего не может → синтаксис и строгий режим.

## 1. Что такое JavaScript

**JavaScript** — объектно-ориентированный скриптовый язык программирования. Он выполняется в браузере (на клиенте) и на сервере (через Node.js).

**Движки JavaScript:**

1. V8 (Chrome, Node.js)
2. SpiderMonkey
3. Chakra
4. Rhino
5. JavaScriptCore

## 2. Браузер как среда выполнения

Браузер (клиент) включает:

- сетевые компоненты;
- движок рендера HTML и CSS;
- движок для обработки скриптов JS;
- движки для изображений, видео и прочих документов;
- компоненты для отладки;
- прочие компоненты.

## 3. Структура клиентского JavaScript

1. **Ядро — ECMAScript** — типы данных, инструкции, ключевые и зарезервированные слова.
2. **BOM** — объектная модель браузера.
3. **DOM** — объектная модель документа.

## 4. Возможности JavaScript

**Может:**

- управлять тегами: создавать, удалять, менять свойства и стили;
- взаимодействовать с пользователем и реагировать на события;
- отправлять запросы на сервер и подгружать данные;
- работать с Cookie, localStorage, IndexedDB.

**Не может:**

1. получать произвольный доступ к файловой системе;
2. управлять файлами и базами данных;
3. получать доступ к соседним вкладкам и окнам пользователя.

**Особенности:** интеграция с HTML и CSS, простые вещи делаются просто, поддерживается всеми браузерами.

## 5. Языки поверх JavaScript

1. CoffeeScript
2. TypeScript

4. Flow

## 6. Синтаксис

Си-подобный синтаксис:

1. чувствительность к регистру букв;
2. строковые литералы в `'одинарных'` и `"двойных"` кавычках;
3. однострочные `//` и многострочные `/* ... */` комментарии;
4. группировка выражений в `(круглых скобках)`.

Пример:

```javascript
const numbers = [1, 2, 3];
const doubled = numbers.map(n => n * 2);
console.log(doubled); // [2, 4, 6]
```

## 7. Строгий режим `'use strict'`

- Включает строгий режим: многие старые невидимые ошибки становятся видимыми в консоли браузера.
- Директива должна быть самой первой в скрипте или в функции.

```javascript
'use strict';
// your code ...
```
